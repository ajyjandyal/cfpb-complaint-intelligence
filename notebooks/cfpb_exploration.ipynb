```python
import requests, pandas as pd

url = "https://www.consumerfinance.gov/data-research/consumer-complaints/search/api/v1/"
params = {"product": "Checking or savings account", "date_received_min": "2026-09-01",
          "date_received_max": "2026-10-01", "format": "csv"}
r = requests.get(url, params=params, headers={"User-Agent": "anku-mba-project/1.0"}, timeout=300)
open("complaints_sep2026.csv", "wb").write(r.content)
df = pd.read_csv("complaints_sep2026.csv")
print(df.shape)
print(df.columns.tolist())
```
